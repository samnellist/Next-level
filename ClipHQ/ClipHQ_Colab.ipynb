# 🎬 Clip HQ (Google Colab version)
Turns long streams into vertical Shorts with captions, hooks and scores. **Runs on Google's computers**, so nothing is installed on your laptop.

**How to use it**
1. Put your videos in **Google Drive → My Drive → ClipHQ → input** (from your iPhone's Google Drive app, or drive.google.com).
2. Press the ▶ button on **Step 1** (do this every time you open this page).
3. Press ▶ on **Step 3** to make clips. They appear in **My Drive → ClipHQ → clips**.
4. On your iPhone: Google Drive app → open a clip → **⋮ → Send a copy → Save Video** to put it in your camera roll.

Tip: **Runtime → Change runtime type → T4 GPU** makes it much faster (free, when Google has one available).


```python
#@title ▶ Step 1: Start Clip HQ  (run this first, every time)
import os, sys, subprocess, zipfile, shutil, pathlib
from google.colab import drive
drive.mount('/content/drive')
APP = '/content/ClipHQ'
BRANCH = 'claude/clip-hq-ai-clipping-tll5ek'
DRIVE_HOME = '/content/drive/MyDrive/ClipHQ'
os.makedirs(DRIVE_HOME + '/input', exist_ok=True)

def _unpack(zip_path):
    with zipfile.ZipFile(zip_path) as z:
        z.extractall('/content/_app')
    for p in pathlib.Path('/content/_app').rglob('colab.py'):
        if p.parent.name == 'cliphq':
            shutil.copytree(p.parent.parent, APP, dirs_exist_ok=True)
            return True
    return False

if not os.path.exists(APP + '/cliphq/colab.py'):
    print('Getting Clip HQ...')
    ok = False
    r = subprocess.run(['wget', '-q', '-O', '/content/app.zip',
                        f'https://codeload.github.com/samnellist/Next-level/zip/refs/heads/{BRANCH}'])
    if r.returncode == 0 and os.path.getsize('/content/app.zip') > 10000:
        ok = _unpack('/content/app.zip')
    if not ok and os.path.exists(DRIVE_HOME + '/ClipHQ.zip'):
        ok = _unpack(DRIVE_HOME + '/ClipHQ.zip')
    if not ok:
        raise SystemExit('Could not get Clip HQ. Upload ClipHQ.zip into My Drive > ClipHQ, then run Step 1 again.')

print('Installing (about 1 minute)...')
subprocess.run('pip install -q faster-whisper httpx cryptography && '
               'pip install -q --no-deps rapidocr-onnxruntime pyclipper', shell=True, check=True)
if APP not in sys.path:
    sys.path.insert(0, APP)
os.chdir(APP)
from cliphq import colab
colab.setup(DRIVE_HOME)
print('\n✅ Ready! Now go to Step 3 (or Step 2 to add a campaign).')
```

```python
#@title Step 2 (optional): Save a campaign
name = "Lacy"  #@param {type:"string"}
hashtags = "#Lacy #shorts"  #@param {type:"string"}
tags = "lacy, gta 6"  #@param {type:"string"}
pay_rate = ""  #@param {type:"string"}
rules = "#Lacy in caption | Lacy must be the main focus | No logos"  #@param {type:"string"}
main_focus = "Lacy"  #@param {type:"string"}
#@markdown Separate rules with a `|`. Running this again with the same name updates the campaign.
colab.add_campaign(name, hashtags, tags, pay_rate, rules, main_focus)
```
